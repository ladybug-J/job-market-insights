````python
json_schema = {
    "role_description": "string",
    "seniority_level": "string",
    "tools": {
        "required": ["string"],
        "beneficial": ["string"]
    },
    "experience": {
        "years": "int",
        "required_knowledge": ["string"] 
    },
    "education": ["string"]        
    "languages": {
        "required": ["string"],
        "beneficial": ["string"]
    }
    "soft_skills": ["string"]    
    }
}
````

````python
def extract_json(description_text):
    import json
    
    system_prompt = """You are an expert in extracting structured data from job descriptions. 
                     When extracting structured data ensure:
                     1. Use standard non-abbreviated words (e.g. 'machine learning' instead of 'ML') 
                     2. Use consistent job descriptions mentioning the position and the subfield or department (e.g. 'data scientist specialized in bioinformatics') 
                     3. List of summarized soft-skills (e.g. ['independent-working', 'teamwork'] instead of 'capacity to work both independently and as part of a team') 
                     4. Return exact JSON format as specified schema and words without capitalization
                     
                     The JSON willl contain the following fields:
                     -'role_description': position name and the department or subfield of study
                     -'seniority_level'
                
                Examples of standardized outputs:
                
                Job description: Looking for a ML Engineer to help decision making in the sales department...
                
                Extracted output:
                
                {
                    'role_description': 'machine learning engineer in sales',
                    'seniority_level': 'not specified',
                    'tools': {
                        'required': ['python', 'structured query language' 'relational databases'],
                        'beneficial': ['amazon web services']
                        }
                    'experience': {
                        'years': 2,
                        'required_knowledge': ['marketing', 'agile development', 'data analysis']
                        },
                    'education': ['computer science', 'data science','physics'],
                    'languages': {
                        'required': ['english', 'spanish'],
                        'beneficial: ['german']
                        },
                    'soft_skills': ['teamwork', 'proactive', 'problem-solving']
                
                }
                    """
    
    user_prompt = f"Extract structured data from the following description:\n\n{description_text}"
    
    output = llm.create_chat_completion(
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        response_format = {
            "type": "json_object",
            "schema": json_schema
        },
        temperature = 0.0,
        max_tokens = 1024
    )
    
    return output["choices"][0]["message"]["content"]
````
